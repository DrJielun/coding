### 3️⃣ IF-ELSE: Making Decisions (Two Choices)

Programs need to make decisions based on conditions<br>

if-else gives you TWO possible paths: true or false <br>

**Diagram:**
```
┌─────────────┐
│  Condition  │
│  (True?)    │
└──────┬──────┘
       │
    Yes│  No
       │   └────> Do nothing
       ↓
  ┌─────────┐
  │ Execute │
  │  Code   │
  └─────────┘
```

**Key Rules:**
1. Start with `if` keyword
2. Add a condition
3. End with a colon `:` 
4. Indent the code inside (use Tab or 4 spaces)

````python
temperature = float(input("What's the temperature today? "))

if temperature > 30:  # Condition: Is temp greater than 30?
    print("It's HOT! Stay hydrated! 🌞")  # Runs if TRUE
else:
    print("Weather is comfortable! 😊")  # Runs if FALSE

# The colon (:) and indentation (4 spaces) are REQUIRED!
````

````python
age = 16  # Create a variable called age and store the value 16

if age >= 18:  # Check if age is greater than or equal to 18
    print("You can get a driving license!")  # Runs if age >= 18 is True
else:  # This means "otherwise" or "if the above condition is False"
    print("You're too young to drive.")  # Runs if age >= 18 is False
````

### 4️⃣ IF-ELIF-ELSE: Making Multiple Decisions

When you have MORE than two possible outcomes<br>
elif = "else if" (checks another condition)<br>

**Diagram:**
```
┌──────────────┐
│ If (first)   │───Yes──> Action 1
└──────┬───────┘
       │No
       ↓
┌──────────────┐
│ Elif (second)│───Yes──> Action 2
└──────┬───────┘
       │No
       ↓
┌──────────────┐
│ Elif (third) │───Yes──> Action 3
└──────┬───────┘
       │No
       ↓
┌──────────────┐
│ Else (last)  │───────> Action 4
└──────────────┘

````python
# Grade Checker
score = 85  # Create a variable to store the student's score

if score >= 90:  # First check: Is score 90 or above?
    grade = "A - Excellent!"  # If yes, assign grade A
elif score >= 80:  # Second check: Is score 80-89? (only checks if first was False)
    grade = "B - Good job!"  # If yes, assign grade B
elif score >= 70:  # Third check: Is score 70-79?
    grade = "C - Fair"  # If yes, assign grade C
elif score >= 60:  # Fourth check: Is score 60-69?
    grade = "D - Need improvement"  # If yes, assign grade D
else:  # If all conditions above are False (score is below 60)
    grade = "F - Failed"  # Assign grade F

print(f"Your score is {score}")  # Display the score using f-string
print(f"Your grade is {grade}")  # Display the grade


# Python checks from TOP to BOTTOM and stops at the FIRST true condition
````

````python
grade = int(input("Enter your test score (0-100): "))

if grade >= 75:  # First condition
    print("Excellent! You passed with distinction! 🌟")
elif grade >= 50:  # Second condition (only checked if first is false)
    print("Good job! You passed! ✅")
elif grade >= 40:  # Third condition
    print("You passed, but need improvement. 📚")
else:  # If ALL conditions are false
    print("Unfortunately, you need to retake the test. 💪")

# Python checks from TOP to BOTTOM and stops at the FIRST true condition
````

````python
# ATM Withdrawal System
balance = 500  # Starting balance in the account (in dollars)
withdrawal = 600  # Amount the customer wants to withdraw

if withdrawal > balance:  # Check if withdrawal amount is MORE than balance
    print("Insufficient funds!")  # Not enough money in account
    print(f"Your balance is ${balance}")  # Show current balance
elif withdrawal <= 0:  # Check if withdrawal is zero or negative
    print("Invalid amount!")  # Can't withdraw zero or negative amount
else:  # If withdrawal is valid and sufficient funds exist
    balance = balance - withdrawal  # Subtract withdrawal from balance
    print(f"Withdrawal successful!")  # Transaction complete
    print(f"New balance: ${balance}")  # Show remaining balance
````
